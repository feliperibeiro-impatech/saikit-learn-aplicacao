```python
import pandas as pd
import numpy as np
from saikit_learn.preprocessing import StandardScaler

df = pd.read_csv("../../../data/processed/superconductivity_duplicated.csv")
df.shape
```

### Pré Processamento dos Dados para Regressão

O dataset de supercondutividade, previamente tratado na etapa de EDA, foi dividido na proporção de 70/30 para, respectivamente, Treino e Teste. 

Foi utilizado a função StandardScaler da nossa biblioteca saikit_learn para padronizar os preditores. O ajuste foi exclusivamente realizado no conjunto de treino, evitando vazamento de informações estatísticas dfo conjunto de teste.

Os conjuntos foram salvos sem padronização, permitindo a aplicação de transformações específicas durante o treinamento dos diferentes m,odelos. 

```python
df_treino = df.sample(frac=0.7, random_state=42)
df_teste = df.drop(df_treino.index)

print("Treino:", df_treino.shape)
print("Teste:", df_teste.shape)
```

```python
X_train = df_treino.drop(columns=["critical_temp"]).to_numpy()
y_train = df_treino["critical_temp"].to_numpy()

X_test = df_teste.drop(columns=["critical_temp"]).to_numpy()
y_test = df_teste["critical_temp"].to_numpy()
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Médias:", X_train_scaled.mean(axis=0)[:5])
print("Desvios:", X_train_scaled.std(axis=0)[:5])
```

```python
df_treino.to_csv("../../../data/processed/treino_regressao.csv", index=False)

df_teste.to_csv("../../../data/processed/teste_regressao.csv", index=False)
```
