# **Pré-processamento — MAGIC Gamma**

## **Objetivo**

Realizar uma análise exploratória do conjunto de dados MAGIC Gamma Telescope, identificando características, distribuições, correlações e possíveis padrões nos dados.

O objetivo é compreender o conjunto de dados antes da aplicação dos algoritmos de classificação.

```python
from saikit_learn.preprocessing import StandardScaler
import pandas as pd
import numpy as np
```

```python
columns = ["fLength", "fWidth", "fSize", "fConc", "fConc1", "fAsym", "fM3Long", "fM3Trans", "fAlpha", "fDist", "class"]

df = pd.read_csv("../../../data/raw/magic04.data", header=None, names=columns)

X = df.drop(columns=["class"])
y = df["class"].map({"g": 1,"h": 0})
```

```python
n = len(X)
indices = np.random.RandomState(42).permutation(n)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

train_indices = indices[:train_end]
val_indices = indices[train_end:val_end]
test_indices = indices[val_end:]

X_train = X.iloc[train_indices]
y_train = y.iloc[train_indices]

X_val = X.iloc[val_indices]
y_val = y.iloc[val_indices]

X_test = X.iloc[test_indices]
y_test = y.iloc[test_indices]
```

## **Verificação da padronização**

Após aplicar o StandardScaler, verificamos que as médias
das variáveis do conjunto de treinamento são aproximadamente
iguais a zero e seus desvios-padrão são iguais a um.

Esses resultados indicam que a transformação foi realizada
corretamente.

Os parâmetros calculados no conjunto de treinamento foram
reutilizados nos conjuntos de validação e teste, evitando
vazamento de dados (data leakage).

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_val_scaled = scaler.transform(X_val)

X_test_scaled = scaler.transform(X_test)
```

```python
print("Média do treinamento:")
print(X_train_scaled.mean(axis=0))

print("\nDesvio padrão do treinamento:")
print(X_train_scaled.std(axis=0))
```

## **Analisando as dimensões dos grupos**

Ao avaliar a divisão dos dados, observamos que as dimensões dos conjuntos correspondem às proporções previamente estabelecidas, mantendo quantidades semelhantes de amostras .

Além disso, as proporções das classes Gamma e Hadron permaneceram próximas entre os conjuntos de treinamento, validação e teste, sem a ausência de elementos inválidos em algum grupo


Essa consistência contribui para que os conjuntos sejam representativos da distribuição original dos dados, favorecendo uma avaliação mais confiável dos modelos de classificação.

```python
print("Treinamento:", X_train_scaled.shape, y_train.shape)
print("Validação:", X_val_scaled.shape, y_val.shape)
print("Teste:", X_test_scaled.shape, y_test.shape)
```

```python
print("Treinamento:")
print(y_train.value_counts(normalize=True) * 100)

print("\nValidação:")
print(y_val.value_counts(normalize=True) * 100)

print("\nTeste:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
for nome, conjunto in [
    ("Treinamento", X_train_scaled),
    ("Validação", X_val_scaled),
    ("Teste", X_test_scaled)
]:
    print(nome)
    print("NaN:", np.isnan(conjunto).sum())
    print("Infinitos:", np.isinf(conjunto).sum())
    print()
```
